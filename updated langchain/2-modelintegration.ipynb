####  Groq model integration

```python
import os 
from langchain_groq import ChatGroq

model=ChatGroq(model="llama-3.3-70b-versatile")
response=model.invoke("why parrots talk?")
response.content
```

#### Streaming 

```python
model.invoke("write me  a  200  words paragraph on artifical Intelligence?")
```

```python
for chunk in model.stream("write me a  200 words paragraph on artifical Intelligence"):
    print(chunk.text,end="|", flush=True)
```

```python
for chunk in model.stream("why do parrots have colorful feathers"):
    print(chunk.text,end="|",flush=True)
```

#### Batch  


```python
response=model.batch([
    "why do parrots have colorful feathers?",
    "how do airplane fly?",
    "what is quatum computing?"
])
for res in response:
    print(res.content)
```
